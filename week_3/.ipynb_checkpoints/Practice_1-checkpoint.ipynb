```python
def printme(str):
    """This is a print function"""
    print(str)
    return

# Calling the function twice
printme("I'm first call to user defined function!")
printme("Again second call to the same function")
```

```python
def changeme(mylist):
    """This changes a passed List"""
    mylist.append([1, 2, 3, 4])  # Modifying the original list
    print("Values inside the function: ", mylist)
    return

# Creating a list and passing to function
mylist = [10, 20, 30]
changeme(mylist)
print("Values outside the function: ", mylist)
```

```python
def changeme(mylist):
    """This creates a new list instead of modifying the passed one"""
    mylist = [1, 2, 3, 4]  # This creates a NEW local list
    print("Values inside the function: ", mylist)
    return

mylist = [10, 20, 30]
changeme(mylist)
print("Values outside the function: ", mylist)
```

```python
def printme(str):
    """This prints a passed string"""
    print(str)
    return
printme();
```

```python
def printinfo(name, age):
    """Test function"""
    print("Name: ", name)
    print("Age: ", age)
    return
printinfo(age=50, name="miki")
```

```python
def printinfo(name, age=35):
    """Test function with default argument"""
    print("Name: ", name)
    print("Age: ", age)
    return
printinfo(name="miki")  # Age defaults to 35
printinfo(age=50, name="miki")  # Age becomes 50
```

```python
def printinfo(arg1, *vartuple):
    """Function with variable-length arguments"""
    print("Output is: ")
    print(arg1)
    for var in vartuple:
        print(var)
    return
printinfo(10)
printinfo(70, 60, 50)

```

```python
# Global variable
total = 50
def sum(arg1, arg2):
    # Local variable (shadows global)
    total = arg1 + arg2
    print("Inside the function local total: ", total)
    return total
# Call the function
sum(10, 20)
print("Outside the function global total: ", total)
```
