```python
import tkinter as tk
import csv

# Function to verify employee
def verify_employee():

    name = name_entry.get().strip()
    department = dept_entry.get().strip()

    found = False
    dept_members = []

    # Open the CSV dataset
    with open("GIG-logistics.csv", newline="") as file:
        reader = csv.DictReader(file)

        for row in reader:

            # Check if employee exists
            if row["Name"].lower() == name.lower() and row["Department"].lower() == department.lower():
                found = True

            # Collect members of the same department
            if row["Department"].lower() == department.lower():
                dept_members.append(row["Name"])

    if found:

        message = f"Welcome {name}!\n\nMembers of {department} Department:\n"

        for member in dept_members:
            if member.lower() != name.lower():
                message += member + "\n"

        result_label.config(text=message)

    else:
        result_label.config(text="Employee does not exist.")


# Create main window
window = tk.Tk()
window.title("GIG Logistics Employee Verification")
window.geometry("450x300")

# Name label and entry
name_label = tk.Label(window, text="Enter Employee Name")
name_label.pack()

name_entry = tk.Entry(window)
name_entry.pack()

# Department label and entry
dept_label = tk.Label(window, text="Enter Department")
dept_label.pack()

dept_entry = tk.Entry(window)
dept_entry.pack()

# Verify button
verify_button = tk.Button(window, text="Verify Employee", command=verify_employee)
verify_button.pack(pady=10)

# Result display
result_label = tk.Label(window, text="", justify="left")
result_label.pack()

# Start GUI loop
window.mainloop()
```
